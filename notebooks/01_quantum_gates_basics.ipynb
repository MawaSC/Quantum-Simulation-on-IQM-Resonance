```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

from qiskit.visualization import plot_histogram, circuit_drawer
import matplotlib.pyplot as plt
```

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.draw("mpl")
```

```python
qc.measure_all()
```

```python
simulator = AerSimulator()
```

```python
result = simulator.run(qc, shots=1000).result()

counts = result.get_counts()

counts

qc.draw("mpl")
```

Apply X, Y, Z, and CNOT Gates

Initialize a qubit in |0>

```python
qc_new = QuantumCircuit(1)
qc_new.draw("mpl")
```

Apply X Gate

```python
qc_new.x(0)
qc_new.draw("mpl")
```

```python
qc_new.measure_all()

simulator = AerSimulator()

result = simulator.run(qc_new, shots=1000).result()

counts = result.get_counts()

counts
```

Apply Y Gate

```python
qc_new = QuantumCircuit(1)

qc_new.y(0)

qc_new.draw("mpl")

qc_new.measure_all()

result = simulator.run(qc_new, shots=1000).result()

counts = result.get_counts()

counts
```

Apply Z Gate

```python
qc_new = QuantumCircuit(1)

qc_new.z(0)

qc_new.draw("mpl")
```

Apply CNOT Gate

```python
qc_new = QuantumCircuit(2)

qc_new.cx(0,1)

qc_new.draw("mpl")
```

```python
qc_new = QuantumCircuit(2)

qc_new.x(0)

qc_new.cx(0,1)

qc_new.draw("mpl")
```

```python
qc_new.measure_all()

result = simulator.run(qc_new, shots=1000).result()

counts = result.get_counts()

counts
```
